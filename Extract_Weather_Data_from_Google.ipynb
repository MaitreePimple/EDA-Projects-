```python
import requests
from bs4 import BeautifulSoup

# 1. Enter city name
city = "South Korea"

# 2. Add a User-Agent header (CRITICAL STEP)
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"
}

# 3. Create URL and make request
url = "https://www.google.com/search?q=" + "weather+" + city
response = requests.get(url, headers=headers)
html = response.content

# 4. Parse with BeautifulSoup
soup = BeautifulSoup(html, 'html.parser')

try:
    # Extracting data using common Google Weather classes
    # Note: These IDs ('wob_tm', etc.) are more stable than the classes in your original code
    temp = soup.find(id='wob_tm').text + "°C"
    sky = soup.find(id='wob_dc').text
    time = soup.find(id='wob_dts').text

    # Extracting Wind and Humidity
    humidity = soup.find(id='wob_hm').text
    wind = soup.find(id='wob_ws').text

    # Printing the output
    print(f"Temperature is: {temp}")
    print(f"Time: {time}")
    print(f"Sky Description: {sky}")
    print(f"Humidity: {humidity}, Wind: {wind}")

except AttributeError:
    print("Error: Google changed its layout. Could not find weather elements.")
```
